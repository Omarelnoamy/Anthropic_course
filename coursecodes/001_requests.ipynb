```python
#Create an API Client 
from anthropic import Anthropic
client = Anthropic()
model = "claude-sonnet-4-6"
```

```python
#Make a request 
message = client.messages.create(
    model= model,
    max_tokens= 1000,
    messages=[
        {
            "role": "user",
            "content": "what is quantum computing? answer in one sentence"
        }
    ]
)
```

```python
def add_user_messages(messages,text):
    user_message= {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_messages(messages,text):
    assistant_message= {"role": "assistant", "content": text}
    messages.append(assistant_message)


#Make a real chat
def chat(messages): 
    message = client.messages.create(
        model= model,
        max_tokens= 1000,
        messages= messages
    )
    return message.content[0].text
```

```python
#Making a starting list of messages
messages =[] 

#Add an initial user message
add_user_messages(messages, "Define Quantum computing in one sentence")

#pass the list of messages to the chat
answer= chat(messages)

#take answer add it in the assistant messages
add_assistant_messages(messages, answer)

#Add in user follow up message
add_user_messages(messages, "Write another sentence")

#call chat again with the list of messages to get the final answer
answer= chat(messages)
answer
```

## Chat Exercise

```python
messages=[]

while True:
    user_input = input("> ")
    print("You: ", user_input)
    add_user_messages(messages, user_input)
    answer = chat(messages)
    add_assistant_messages(messages, answer) 
    print("---\n", "AI: ",answer, "\n---")   
```
